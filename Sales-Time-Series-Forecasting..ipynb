**Time Series Data Preparation**

```python

# 1. Convert Date column into datetime
df['Date'] = pd.to_datetime(df['Date'])

# 2. Sort data by date
df = df.sort_values(by='Date')

# 3. Set Date as index
df = df.set_index('Date')

# 4. Display first five records
print("TIME SERIES DATA:")
display(df.head())

# 5. Display last five records
print("\nLATEST RECORDS:")
display(df.tail())

# 6. Check date range
print("\nDATE RANGE:")
print("Start Date:", df.index.min())
print("End Date:", df.index.max())

# 7. Check frequency
print("\nDATE FREQUENCY:")
print(df.index.to_series().diff().value_counts().head())

print("\nTIME SERIES PREPARATION COMPLETED!")
```

**Sales Trend Analysis and Visualization (EDA)**

**Time Series Decomposition**

```python
from statsmodels.tsa.seasonal import seasonal_decompose

# Aggregate sales by date
daily_sales = df.groupby(df.index)['Units Sold'].sum().sort_index()

# Create continuous daily dates
daily_sales = daily_sales.asfreq('D')

# Fill missing dates
daily_sales = daily_sales.interpolate(method='time')

# Perform decomposition
decomposition = seasonal_decompose(
    daily_sales,
    model='additive',
    period=7
)

# Plot decomposition
fig = decomposition.plot()
fig.set_size_inches(14, 10)

plt.suptitle("Time Series Decomposition of Daily Sales", fontsize=16)
plt.tight_layout()
plt.show()

# Display components
print("TREND COMPONENT:")
display(decomposition.trend.dropna().head())

print("\nSEASONALITY COMPONENT:")
display(decomposition.seasonal.head(14))

print("\nRESIDUAL COMPONENT:")
display(decomposition.resid.dropna().head())

print("\nTIME SERIES DECOMPOSITION COMPLETED!")
```

**Stationarity Testing (ADF Test)**

```python
from statsmodels.tsa.stattools import adfuller

# Prepare daily sales
daily_sales = df.groupby(df.index)['Units Sold'].sum().sort_index()

# Run Augmented Dickey-Fuller test
adf_result = adfuller(daily_sales.dropna())

# Display results
print("AUGMENTED DICKEY-FULLER (ADF) TEST")
print("-----------------------------------")

print("ADF Statistic:", adf_result[0])
print("p-value:", adf_result[1])
print("Number of Lags:", adf_result[2])
print("Number of Observations:", adf_result[3])

print("\nCRITICAL VALUES:")
for key, value in adf_result[4].items():
    print(key, ":", value)

# Interpret results
print("\nSTATIONARITY RESULT:")

if adf_result[1] <= 0.05:
    print("The sales data is STATIONARY.")
else:
    print("The sales data is NON-STATIONARY.")
    print("Differencing may be required before ARIMA modelling.")

print("\nADF TEST COMPLETED!")
```

**Train-Test Split**

```python
# Aggregate daily sales
daily_sales = df.groupby(df.index)['Units Sold'].sum().sort_index()

# Calculate 80% training size
train_size = int(len(daily_sales) * 0.80)

# Split the data chronologically
train = daily_sales.iloc[:train_size]
test = daily_sales.iloc[train_size:]

# Display dataset sizes
print("TRAINING AND TESTING DATA")
print("-------------------------")

print("Total Records:", len(daily_sales))
print("Training Records:", len(train))
print("Testing Records:", len(test))

print("\nTRAINING DATE RANGE:")
print("Start:", train.index.min())
print("End:", train.index.max())

print("\nTESTING DATE RANGE:")
print("Start:", test.index.min())
print("End:", test.index.max())

# Visualize the split
plt.figure(figsize=(15, 5))

plt.plot(train.index, train, label="Training Data")
plt.plot(test.index, test, label="Testing Data")

plt.title("Sales Data: Training vs Testing")
plt.xlabel("Date")
plt.ylabel("Units Sold")
plt.legend()
plt.grid(True)
plt.show()

print("\nTRAIN-TEST SPLIT COMPLETED!")
```

**Build a Baseline Forecasting Model (Moving Average)**

```python
# Calculate the average sales of the previous 7 days
window_size = 7

# Store actual test sales
actual_sales = test.copy()

# Create predictions
history = list(train.values)
predictions = []

for i in range(len(test)):

    # Calculate the average of the previous 7 days
    prediction = np.mean(history[-window_size:])

    # Save prediction
    predictions.append(prediction)

    # Add actual sales to history
    history.append(test.iloc[i])

# Convert predictions into a Pandas Series
predictions = pd.Series(
    predictions,
    index=test.index,
    name="Predicted Sales"
)

# Display actual vs predicted values
results = pd.DataFrame({
    "Actual Sales": actual_sales,
    "Predicted Sales": predictions
})

print("MOVING AVERAGE FORECAST RESULTS")
display(results.head(10))

# Visualize actual vs predicted sales
plt.figure(figsize=(15, 5))

plt.plot(train.index, train, label="Training Sales")
plt.plot(actual_sales.index, actual_sales, label="Actual Sales")
plt.plot(predictions.index, predictions, label="Predicted Sales")

plt.title("Moving Average Sales Forecast")
plt.xlabel("Date")
plt.ylabel("Units Sold")
plt.legend()
plt.grid(True)
plt.show()

print("MOVING AVERAGE MODEL COMPLETED!")
```

**Model Evaluation (MAE, RMSE and MAPE)**

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_percentage_error

# Calculate evaluation metrics
mae = mean_absolute_error(actual_sales, predictions)

rmse = np.sqrt(
    mean_squared_error(actual_sales, predictions)
)

mape = mean_absolute_percentage_error(
    actual_sales,
    predictions
) * 100

# Display results
print("MODEL EVALUATION RESULTS")
print("------------------------")

print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("MAPE:", round(mape, 2), "%")

# Display evaluation summary
evaluation_results = pd.DataFrame({
    "Metric": ["MAE", "RMSE", "MAPE"],
    "Value": [mae, rmse, mape]
})

display(evaluation_results)

print("MODEL EVALUATION COMPLETED!")
```

**Build ARIMA Forecasting Model**

```python
from statsmodels.tsa.arima.model import ARIMA

# Select differencing order using ADF test result
if adf_result[1] <= 0.05:
    d = 0
else:
    d = 1

print("ARIMA PARAMETERS")
print("----------------")
print("p = 1")
print("d =", d)
print("q = 1")

# Build ARIMA model
arima_model = ARIMA(
    train,
    order=(1, d, 1)
)

# Train the model
arima_fit = arima_model.fit()

# Forecast test-period sales
arima_forecast = arima_fit.forecast(
    steps=len(test)
)

# Convert predictions into a Pandas Series
arima_predictions = pd.Series(
    arima_forecast.values,
    index=test.index,
    name="ARIMA Predicted Sales"
)

# Display forecast results
arima_results = pd.DataFrame({
    "Actual Sales": test,
    "ARIMA Predicted Sales": arima_predictions
})

print("\nARIMA FORECAST RESULTS")
display(arima_results.head(10))

# Visualize actual vs predicted sales
plt.figure(figsize=(15, 5))

plt.plot(train.index, train, label="Training Sales")
plt.plot(test.index, test, label="Actual Sales")
plt.plot(
    arima_predictions.index,
    arima_predictions,
    label="ARIMA Predictions"
)

plt.title("ARIMA Sales Forecast")
plt.xlabel("Date")
plt.ylabel("Units Sold")
plt.legend()
plt.grid(True)
plt.show()

print("\nARIMA MODEL SUMMARY")
print(arima_fit.summary())

print("\nARIMA MODEL TRAINING COMPLETED!")
```

**Evaluate ARIMA Model and Compare It with Moving Average**

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_percentage_error

# Moving Average Evaluation
mae_ma = mean_absolute_error(actual_sales, predictions)

rmse_ma = np.sqrt(
    mean_squared_error(actual_sales, predictions)
)

mape_ma = mean_absolute_percentage_error(
    actual_sales, predictions
) * 100

# ARIMA Evaluation
mae_arima = mean_absolute_error(test, arima_predictions)

rmse_arima = np.sqrt(
    mean_squared_error(test, arima_predictions)
)

mape_arima = mean_absolute_percentage_error(
    test, arima_predictions
) * 100

# Create comparison table
comparison = pd.DataFrame({
    "Model": ["Moving Average", "ARIMA"],
    "MAE": [mae_ma, mae_arima],
    "RMSE": [rmse_ma, rmse_arima],
    "MAPE (%)": [mape_ma, mape_arima]
})

print("MODEL PERFORMANCE COMPARISON")
print("----------------------------")

display(comparison.round(2))

# Identify the model with the lowest MAE
best_model = comparison.loc[
    comparison["MAE"].idxmin(), "Model"
]

print("\nBEST MODEL BASED ON MAE:", best_model)

# Visualize model comparison
plt.figure(figsize=(10, 5))

plt.bar(
    comparison["Model"],
    comparison["MAE"]
)

plt.title("Model Comparison - MAE")
plt.xlabel("Forecasting Model")
plt.ylabel("Mean Absolute Error")
plt.grid(axis="y")
plt.show()

print("\nMODEL COMPARISON COMPLETED!")
```

**Forecast Future Sales for the Next 30 Days.**

```python

# Set forecasting period
forecast_days = 30

# Generate future sales predictions
future_forecast = arima_fit.forecast(
    steps=forecast_days
)

# Create future dates
future_dates = pd.date_range(
    start=daily_sales.index.max() + pd.Timedelta(days=1),
    periods=forecast_days,
    freq='D'
)

# Create forecast DataFrame
future_predictions = pd.DataFrame({
    "Date": future_dates,
    "Predicted Units Sold": future_forecast.values
})

# Display future predictions
print("NEXT 30 DAYS SALES FORECAST")
print("----------------------------")

display(future_predictions)

# Visualize future sales
plt.figure(figsize=(15, 5))

plt.plot(
    daily_sales.index,
    daily_sales.values,
    label="Historical Sales"
)

plt.plot(
    future_dates,
    future_forecast.values,
    label="Future Sales Forecast",
    marker='o'
)

plt.title("Next 30 Days Sales Forecast")
plt.xlabel("Date")
plt.ylabel("Units Sold")
plt.legend()
plt.grid(True)
plt.show()

# Save forecast results
future_predictions.to_csv(
    "future_sales_forecast.csv",
    index=False
)

print("\nForecast saved successfully!")
print("FUTURE SALES FORECAST COMPLETED!")
```
